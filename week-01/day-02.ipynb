```python
import requests
from dotenv import load_dotenv
import os
from openai import OpenAI
```

```python
# checking if ollama is running locally
requests.get("http://localhost:11434").content
```

```python
load_dotenv(override=True)
api_url = os.getenv("OLLAMA_BASE_URL")
```

```python
ollama = OpenAI(base_url=api_url, api_key="ollama")
```

```python
user_message=""
print("Welcome Translator")
language = input("Enter language to translate to:")
print(f"\nTranslating from English to {language}\n")
while user_message != "exit":
    sentence = input(f"Enter text to be translated to {language}:").lower()

    if sentence == "exit":
        print("Exited Translator")
        break

    response = ollama.chat.completions.create(model="llama3.2", messages=[{"role": "user", "content": f"Translate the following sentence from English to {language.capitalize()}: {sentence}. The output provided should contain only the translated sentence, nothing else"}])
    print(f"Translation for '{sentence}': '{response.choices[0].message.content}'")
    user_message = sentence
```
